```python
import pandas as pd
import matplotlib.pyplot as plt
from wordcloud import WordCloud

df = pd.read_csv('../data/processed/cleaned_reviews.csv')

df['sentiment'].value_counts().plot(kind='bar')
plt.savefig('../outputs/figures/class_distribution.png')
plt.show()

pos_text = ' '.join(df[df['sentiment']==1]['clean_text'])
neg_text = ' '.join(df[df['sentiment']==0]['clean_text'])

WordCloud().generate(pos_text).to_file('../outputs/figures/pos_wc.png')
WordCloud().generate(neg_text).to_file('../outputs/figures/neg_wc.png')
```
